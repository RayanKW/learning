Решение

```python
%%time
import numpy as np
import numpy.typing as npt


def count(array: npt.NDArray[np.int64], n: np.int64) -> np.uint64:
    x = np.full(array.size, n, dtype=np.int64)
    return (array == x).astype(np.uint64).sum()


count(np.array([7, 9, 3, 7, 9, 3, 4, 7, 8]), 7)
```

```python
%%time
import numpy as np
import numpy.typing as npt


def count(array: npt.NDArray[np.int64], n: np.int64) -> np.uint64:
    return array[array == n].size


count(np.array([7, 9, 3, 7, 9, 3, 4, 7, 8]), 7)
```
